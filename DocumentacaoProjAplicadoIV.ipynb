Como as observações do Novo CAGED são disponibilizadas mensalmente e apresentam dependência temporal, os dados podem ser organizados como uma série temporal. Segundo Hyndman e Athanasopoulos (2021), a construção de modelos de previsão deve considerar métodos simples de referência como forma de comparação. Para séries mensais que apresentam comportamento sazonal, uma alternativa é utilizar como previsão o valor observado no mesmo mês do ano anterior. Além disso, a avaliação deve respeitar a ordem temporal dos dados, utilizando informações disponíveis no passado para estimar períodos posteriores.

**Estudos Anteriores sobre Emprego Formal**

---


A utilização de séries temporais para investigar o comportamento do emprego formal já foi explorada em trabalhos anteriores. Campos e Campos (2009) utilizaram dados do CAGED e da RAIS, juntamente com procedimentos de ajuste sazonal e modelos ARMA, para analisar os impactos da crise financeira de 2008 sobre o emprego formal no Nordeste brasileiro. O estudo demonstra a possibilidade de utilizar registros administrativos do mercado de trabalho para investigar alterações associadas a períodos de instabilidade econômica.
Andaku e Horie (2018), por sua vez, analisaram a sazonalidade do emprego formal em diferentes setores utilizando o método X-13-ARIMA-SEATS. O trabalho evidencia que o comportamento sazonal pode variar de acordo com a atividade econômica, reforçando a importância de considerar características temporais na análise do emprego formal.

**Modelos de Previsão aplicados ao Emprego**

---


Mais recentemente, Raggi (2023) comparou diferentes métodos de previsão, incluindo Holt-Winters, ARIMA, SARIMA, TBATS e Prophet, aplicados à previsão de empregos na construção civil a partir de dados do CAGED e da RAIS. No estudo analisado, o método Holt-Winters apresentou o menor erro entre as alternativas avaliadas. Esse resultado reforça a importância da comparação empírica entre diferentes modelos, uma vez que métodos mais complexos não necessariamente apresentam desempenho superior em todos os cenários. O trabalho também aponta como limitação dos modelos univariados a ausência de variáveis externas, aspecto relevante para a interpretação dos resultados obtidos em previsões do mercado de trabalho.

**Estratégia de Modelagem e Avaliação**

---


Com base na literatura analisada, este projeto adotará inicialmente um baseline sazonal como referência e um modelo estatístico de séries temporais, permitindo comparar o desempenho das abordagens de forma consistente. A avaliação será realizada por meio de métricas de erro, como MAE e RMSE, preservando a ordem temporal das observações. Caso seja identificado ganho metodológico, poderá ser avaliada adicionalmente a utilização do XGBoost, incorporando variáveis derivadas da própria série, como defasagens e médias móveis. Entretanto, considerando que a série mensal disponível a partir de 2020 possui quantidade limitada de observações, essa abordagem será tratada como complementar, priorizando modelos mais parcimoniosos, interpretáveis e adequados ao volume de dados disponível.

**Etapas do Pipeline**
**Extração**

Os arquivos mensais do Novo CAGED são obtidos a partir da fonte oficial do MTE. Na carga inicial, é considerada a série histórica disponível desde janeiro de 2020. Em atualizações posteriores, novas competências podem ser incorporadas de forma incremental.

**Camada Bronze — Dados Brutos**

A camada Bronze preserva os arquivos exatamente como disponibilizados pela fonte, organizados por ano e competência. Também devem ser mantidos metadados de ingestão, como data de processamento, fonte e identificação do arquivo, permitindo auditoria e reprocessamento.

**Validação**

Antes do tratamento, os dados passam por verificações de estrutura e qualidade. Entre os controles estão a conferência das colunas esperadas, tipos de dados, competências, valores ausentes, duplicidades, códigos inválidos e consistência das movimentações de admissão e desligamento.

**Camada Silver — Tratamento e Padronização**

Na camada Silver são realizadas a padronização de nomes e tipos, tratamento de datas e campos de movimentação e organização de variáveis necessárias para as análises. O objetivo é produzir uma representação consistente e reutilizável dos registros de movimentação.

**Agregação Temporal**

Os registros tratados são agregados por competência mensal, obtendo-se, principalmente, o total de admissões, o total de desligamentos e o saldo de empregos formais. Essa etapa transforma os microdados em uma série temporal adequada à análise.

**Camada Gold — Base Analítica**

A camada Gold concentra os indicadores mensais consolidados que serão utilizados na análise exploratória, na engenharia de atributos e nos modelos de previsão. Essa camada funciona como a principal entrada das etapas analíticas do projeto.

**Análise Exploratória e Engenharia de Atributos**

A série temporal será analisada quanto à tendência, sazonalidade, alterações de nível e períodos atípicos. Quando aplicável, poderão ser criadas variáveis defasadas, como lags de 1, 3, 6 e 12 meses, médias móveis e variáveis relacionadas ao mês do ano. A construção dos atributos deve respeitar a ordem temporal para evitar vazamento de informação futura.

**Modelagem**

A estratégia de modelagem parte de uma linha de base sazonal, utilizando como referência o comportamento do mesmo período anterior, e de um modelo estatístico de séries temporais. Modelos adicionais, como XGBoost, podem ser considerados caso apresentem ganho metodológico e sejam compatíveis com a quantidade de observações disponíveis.

**Avaliação e Previsão**

A avaliação será realizada preservando a ordem temporal dos dados, utilizando uma separação entre períodos de treinamento e teste. As métricas principais previstas são MAE e RMSE, permitindo comparar o desempenho dos modelos. Após a avaliação, o modelo selecionado poderá ser utilizado para gerar previsões para os meses seguintes, acompanhadas das respectivas limitações.

# **Modelos**


Neste projeto, são utilizadas técnicas de análise e modelagem de séries temporais com o objetivo de compreender o comportamento do saldo mensal de empregos formais no Brasil e desenvolver previsões para períodos futuros. Entre as técnicas empregadas, destacam-se as médias móveis e a suavização exponencial, aplicadas em diferentes etapas do processo analítico.

Médias Móveis (MA): são utilizadas na análise exploratória dos dados para suavizar as oscilações mensais e facilitar a identificação de tendências ao longo do tempo. No projeto, é calculada uma média móvel de 12 meses do saldo de empregos, permitindo observar o comportamento geral da série e reduzir a influência de variações pontuais.

Suavização Exponencial — Holt-Winters é utilizada na etapa de previsão da série temporal. O modelo configurado considera tendência aditiva amortecida e sazonalidade aditiva com período de 12 meses, permitindo representar padrões de evolução e possíveis comportamentos sazonais do saldo de empregos formais.

Além dessas técnicas, o projeto utiliza a decomposição STL (Seasonal and Trend decomposition using Loess) para separar a série temporal em componentes de tendência, sazonalidade e resíduos. Essa abordagem auxilia na identificação de padrões temporais e na investigação de oscilações que não são explicadas pelos componentes estimados. A decomposição é empregada para fins de análise exploratória, não como modelo direto de previsão.

**Especificação dos modelos**

O baseline sazonal repete o saldo do mesmo mês do ano anterior. O Holt-Winters utiliza nível, tendência aditiva amortecida e sazonalidade aditiva de período 12. A formulação aditiva permite saldos negativos. Os parâmetros foram estimados somente no treinamento. A convergência numérica do otimizador não equivale a validação preditiva.

A avaliação dos modelos é realizada por meio das métricas MAE (Mean Absolute Error) e RMSE (Root Mean Squared Error), que quantificam as diferenças entre os valores reais e previstos. O MAE representa o erro absoluto médio, enquanto o RMSE penaliza mais intensamente os erros de maior magnitude.

# **Dicussão e Conclusão**

A implementação produziu uma série mensal contínua e verificações explícitas de qualidade, permitindo analisar o comportamento nacional do emprego formal no recorte MOV. A exploração identificou forte variabilidade, quedas recorrentes em dezembro e dependência anual, além de meses atípicos que foram preservados para representar o fenômeno observado.

A comparação inicial favoreceu o Holt-Winters nas duas métricas, com redução de 10,79% no MAE e 5,49% no RMSE. O resultado constitui evidência parcial de utilidade em relação à repetição do ano anterior. Entretanto, os erros elevados em março de 2026 e dezembro de 2025, bem como a previsão de sinal incorreto em abril de 2026, impedem tratar o modelo como solução plenamente validada.

A principal limitação de cobertura é a exclusão dos arquivos CAGEDFOR e CAGEDEXC. O alvo corresponde ao saldo das movimentações MOV, que pode divergir do saldo oficial ajustado. Os arquivos representam a fotografia disponível na data de coleta, sem reconstrução das versões conhecidas em cada data histórica. Assim, o teste temporal não constitui uma simulação rigorosa de previsão em tempo real.

A agregação nacional oculta diferenças entre unidades da Federação e setores. A série iniciada em 2020 contém poucas repetições anuais e não utiliza variáveis externas. As análises descrevem associações temporais, mas não identificam causas das oscilações. O saldo tampouco permite avaliar diretamente empregabilidade, informalidade ou qualidade dos vínculos.
